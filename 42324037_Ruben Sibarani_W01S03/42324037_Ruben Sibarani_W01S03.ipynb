```python
# Nama: Ruben Sibarani
# NIM: 42324037

import numpy as np
import matplotlib.pyplot as plt


x = np.linspace(-2, 2, 1000)
y = np.linspace(-2, 2, 1000)


X, Y = np.meshgrid(x, y)


F = (X**6
     + 3*X**4*Y**2
     - 3*X**4
     + 3*X**2*Y**4
     - X**2*Y**3
     - 6*X**2*Y**2
     + 3*X**2
     + Y**6
     - 3*Y**4
     + 3*Y**2)


plt.figure(figsize=(8, 8))


plt.contour(X, Y, F, levels=[1])

plt.xlabel("x")
plt.ylabel("y")
plt.title("Grafik 2 Dimensi")

plt.grid(True)

plt.axis("equal")

plt.show()
```

```python
# Nama: Ruben Sibarani
# NIM: 42324037

import numpy as np
import matplotlib.pyplot as plt


x = np.linspace(-5, 5, 300)
y = np.linspace(-5, 5, 300)

X, Y = np.meshgrid(x, y)


Z = ((5 * Y / 4) - np.abs(X))**2 + X**2


fig = plt.figure(figsize=(10, 7))

ax = fig.add_subplot(111, projection="3d")

ax.plot_surface(X, Y, Z)

ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("z")


ax.set_title("Grafik 3 Dimensi")


plt.show()
```
